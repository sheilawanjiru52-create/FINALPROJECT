## Data Cleaning

```python

import pandas as pd

# Identify columns that should contain numeric nutrient values
nutrient_columns = [
    'Energy (kJ)',
    'Energy (kcal)',
    'Water (g)',
    'Protein (g)',
    'Fat (g)',
    'Carbohydrate available (g)',
    'Fibre (g)',
    'Ash (g)',
    'Calcium (mg)',
    'Iron (mg)',
    'Magnesium (mg)',
    'Phosphorus (mg)',
    'Potassium (mg)',
    'Sodium (mg)',
    'Zinc (mg)',
    'Selenium (mcg)',
    'Vitamin A RAE (mcg)',
    'Vitamin A RE (mcg)',
    'Retinol (mcg)',
    'Beta-carotene equivalent (mcg)',
    'Thiamin (mg)',
    'Riboflavin (mg)',
    'Niacin (mg)',
    'Dietary Folate Equivalent (mcg)',
    'Food folate (mcg)',
    'Vitamin B12 (mcg)',
    'Vitamin C (mg)'
]

# Convert nutrient columns to numeric
# Unrecognized text values will become NaN
for col in nutrient_columns:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# Check the updated data types
print(df[nutrient_columns].dtypes)

# Check missing values after conversion
print("\nMissing values after conversion:")
print(df[nutrient_columns].isna().sum().sort_values(ascending=False))
```

```python
original_df = pd.read_csv("food_composition_project.csv")
```

```python
df = original_df.copy()
```

Convert tr entries into missing numeric values (NaN) so Python can perform calculations.

Record their original meaning in separate status columns so the app can still display "Trace amount" instead of "Unknown".

```python
# Nutrient columns that should contain numeric values
nutrient_columns = [
    'Energy (kJ)', 'Energy (kcal)', 'Water (g)', 'Protein (g)',
    'Fat (g)', 'Carbohydrate available (g)', 'Fibre (g)', 'Ash (g)',
    'Calcium (mg)', 'Iron (mg)', 'Magnesium (mg)', 'Phosphorus (mg)',
    'Potassium (mg)', 'Sodium (mg)', 'Zinc (mg)', 'Selenium (mcg)',
    'Vitamin A RAE (mcg)', 'Vitamin A RE (mcg)', 'Retinol (mcg)',
    'Beta-carotene equivalent (mcg)', 'Thiamin (mg)',
    'Riboflavin (mg)', 'Niacin (mg)',
    'Dietary Folate Equivalent (mcg)', 'Food folate (mcg)',
    'Vitamin B12 (mcg)', 'Vitamin C (mg)'
]

for col in nutrient_columns:
    # Read values as text temporarily
    values = df[col].astype("string").str.strip()

    # Create a status column to preserve the meaning
    status_col = f"{col} Status"

    df[status_col] = "known"
    df.loc[values.isna() | values.eq(""), status_col] = "unknown"

    trace_mask = values.str.lower().eq("tr").fillna(False)
    df.loc[trace_mask, status_col] = "trace"

    # Convert nutrient values to numeric
    # Trace and empty values become NaN
    df[col] = pd.to_numeric(
        values.mask(trace_mask | values.eq("")),
        errors="coerce"
    )

    # Flag unexpected nonnumeric entries for review
    invalid_mask = (
        values.notna()
        & values.ne("")
        & df[col].isna()
        & ~trace_mask
    )
    df.loc[invalid_mask, status_col] = "check"

print("Working dataset shape:", df.shape)
print("\nTrace entries by nutrient:")

for col in nutrient_columns:
    count = (df[f"{col} Status"] == "trace").sum()
    if count > 0:
        print(f"{col}: {count}")

print("\nStatus categories:")
print(df[[f"{col} Status" for col in nutrient_columns]]
      .stack().value_counts())
```

Known: A numeric nutrient value is available.

Trace: The source explicitly reports a trace amount.

Unknown: The original cell is empty.

Check: The entry contains something other than a number, tr, or an empty value and needs investigation.

#### Remove square brackets and convert the values

```python
# Clean nutrient values that contain square brackets
for col in nutrient_columns:
    status_col = f"{col} Status"

    # Identify entries that need checking
    check_mask = df[status_col] == "check"

    # Retrieve the original text and remove square brackets
    cleaned_values = (
        original_df.loc[check_mask, col]
        .astype("string")
        .str.replace(r"[\[\]]", "", regex=True)
        .str.strip()
    )

    # Convert the cleaned values to numbers
    df.loc[check_mask, col] = pd.to_numeric(
        cleaned_values, errors="coerce"
    )

    # Update status after successful conversion
    successful = df.loc[check_mask, col].notna()
    successful_indices = df.loc[check_mask].index[successful]

    df.loc[successful_indices, status_col] = "known"

# Check whether any entries still need attention
remaining_checks = sum(
    (df[f"{col} Status"] == "check").sum()
    for col in nutrient_columns
)

print(remaining_checks)
```

## K-means Clustering

#### a.) Feature Selection 

Select all featues that will be used for clustering (All the numerical features)

```python

# Select numerical nutrient features for clustering
clustering_features = [
    'Energy (kcal)',
    'Water (g)',
    'Protein (g)',
    'Fat (g)',
    'Carbohydrate available (g)',
    'Fibre (g)',
    'Calcium (mg)',
    'Iron (mg)',
    'Magnesium (mg)',
    'Phosphorus (mg)',
    'Potassium (mg)',
    'Sodium (mg)',
    'Zinc (mg)',
    'Selenium (mcg)',
    'Vitamin A RAE (mcg)',
    'Thiamin (mg)',
    'Riboflavin (mg)',
    'Niacin (mg)',
    'Vitamin B12 (mcg)',
    'Vitamin C (mg)'
]

# Create the feature DataFrame
X = df[clustering_features].copy()

print("Feature matrix shape:", X.shape)
print("\nMissing values per feature:")
print(X.isna().sum().sort_values(ascending=False))

print("\nPreview of selected features:")
display(X.head())
```

```python
print("Shape of X:", X.shape)

print("\nMissing values in selected features:")
print(X.isnull().sum().sort_values(ascending=False))

print("\nData types:")
print(X.dtypes)
```

Two different rules will be used for this dataset, in preparation for clustering:

- Unknown values: replace them with the median for that nutrient.

- Trace values (tr): replace them with 0 for clustering, treating trace amounts as negligible.

#### b.) Handle unknown and trace values

```python
from sklearn.impute import SimpleImputer

# Work on a separate copy of the selected numerical features
X_cluster = X.copy()

# Replace trace values with zero, using the status columns in df
for col in X_cluster.columns:
    status_col = f"{col} Status"

    if status_col in df.columns:
        trace_mask = df[status_col].eq("trace")
        X_cluster.loc[trace_mask, col] = 0

# Fill remaining unknown values with the median of each nutrient
imputer = SimpleImputer(strategy="median")

X_imputed = pd.DataFrame(
    imputer.fit_transform(X_cluster),
    columns=X_cluster.columns,
    index=X_cluster.index
)

print("Shape:", X_imputed.shape)
print("Missing values remaining:", X_imputed.isnull().sum().sum())

print("\nPreview:")
display(X_imputed.head())
```

#### c.) Scale the features

Standard Scaler - First Approach 

```python
from sklearn.preprocessing import StandardScaler


scaler = StandardScaler()

X_scaled = pd.DataFrame(
    scaler.fit_transform(X_imputed),
    columns=X_imputed.columns,
    index=X_imputed.index
)

print("Scaled data shape:", X_scaled.shape)

print("\nFeature means (should be close to 0):")
print(X_scaled.mean().round(3))

print("\nFeature standard deviations (should be close to 1):")
print(X_scaled.std(ddof=0).round(3))

display(X_scaled.head())
```

```python
print("Type:", type(X_scaled))
print("Shape:", X_scaled.shape)


display(X_scaled.head())

print("\nMeans:")
print(X_scaled.mean().round(3))

print("\nStandard deviations:")
print(X_scaled.std(ddof=0).round(3))
```

```python
# Check for missing values
print("Missing values:", X_scaled.isnull().sum().sum())
```

#### Find K

Elbow Method

```python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# Store the within-cluster sum of squares
inertia = []

# Test different numbers of clusters
K_values = range(1, 11)

for k in K_values:
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)

# Plot the Elbow Curve
plt.figure(figsize=(8, 5))
plt.plot(K_values, inertia, marker='o')
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method to find K")
plt.xticks(K_values)
plt.grid(True)
plt.show()
```

Silhouette Score

```python
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt

silhouette_scores = []
K_values = range(2, 11)

for k in K_values:
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    cluster_labels = kmeans.fit_predict(X_scaled)

    score = silhouette_score(X_scaled, cluster_labels)
    silhouette_scores.append(score)

    print(f"K = {k}, Silhouette Score = {score:.4f}")

# Plot the scores
plt.figure(figsize=(8, 5))
plt.plot(K_values, silhouette_scores, marker="o")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Score for Different K Values")
plt.xticks(list(K_values))
plt.grid(True)
plt.show()
```

**K = 2, score 0.4012: Best result among the tested values. The two groups have reasonably good separation, although there is still some overlap.**

**K = 3–5, scores 0.2949–0.3157: These solutions have weaker average separation.**

**K = 6–10, scores around 0.331–0.339: These perform better than K = 3–5, but not as well as K = 2.**

```python
# Fit the K-Means model
kmeans = KMeans(
    n_clusters=2,
    random_state=42,
    n_init=10
)

cluster_labels = kmeans.fit_predict(X_scaled)

# Keep a copy of the original nutrient data
# df should contain the original, unscaled nutrient values
clustered_df = df.copy()
clustered_df["Cluster"] = cluster_labels

# Compare the average nutrient values for each cluster
cluster_profiles = clustered_df.groupby("Cluster")[
    X_scaled.columns
].mean()

print("Number of foods in each cluster:")
print(clustered_df["Cluster"].value_counts().sort_index())

print("\nAverage nutrient profile by cluster:")
display(cluster_profiles.round(2))
```

## Alternative: Robust Scaling

Uses the median and the interquartile range (IQR), which are less sensitive to extreme values.

```python
from sklearn.preprocessing import RobustScaler

# same numerical nutrient features as before
nutrient_columns = X_scaled.columns

# Apply Robust Scaling to the original nutrient values
robust_scaler = RobustScaler()

X_robust = pd.DataFrame(
    robust_scaler.fit_transform(df[nutrient_columns]),
    columns=nutrient_columns,
    index=df.index
)

print("Robust-scaled dataset shape:", X_robust.shape)
display(X_robust.head())

# Evaluate different numbers of clusters
robust_scores = {}

for k in range(2, 11):
    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(X_robust)
    score = silhouette_score(X_robust, labels)

    robust_scores[k] = score

    print(f"K = {k}: Silhouette score = {score:.4f}")

# Identify the highest silhouette score
best_k = max(robust_scores, key=robust_scores.get)

print(f"\nBest K using RobustScaler: {best_k}")
print(f"Best silhouette score: {robust_scores[best_k]:.4f}")
```

```python
scalers = {
    "StandardScaler": StandardScaler(),
    "RobustScaler": RobustScaler()
}

comparison = {}

for scaler_name, scaler in scalers.items():
    X_scaled_test = scaler.fit_transform(X_zero)

    comparison[scaler_name] = {}

    print(f"\n{scaler_name}")

    for k in range(2, 11):
        model = KMeans(
            n_clusters=k,
            random_state=42,
            n_init=10
        )

        labels = model.fit_predict(X_scaled_test)
        score = silhouette_score(X_scaled_test, labels)

        comparison[scaler_name][k] = score

        print(f"K = {k}: Silhouette score = {score:.4f}")

print("\nBest K for each scaler:")

for scaler_name, scores in comparison.items():
    best_k = max(scores, key=scores.get)
    print(
        f"{scaler_name}: K = {best_k}, "
        f"score = {scores[best_k]:.4f}"
    )
```

```python
from sklearn.cluster import DBSCAN
from sklearn.metrics import (
    silhouette_score,
    davies_bouldin_score,
    calinski_harabasz_score
)
import numpy as np
# Use the same scaled features as the K-Means baseline
X_dbscan = X_scaled

# Fit DBSCAN with initial parameters
dbscan = DBSCAN(
    eps=1.0,
    min_samples=5
)

dbscan_labels = dbscan.fit_predict(X_dbscan)

# Save labels for later inspection
df["DBSCAN_Cluster"] = dbscan_labels

# Count clusters and noise points
unique_labels = set(dbscan_labels)
n_clusters = len(unique_labels - {-1})
n_noise = np.sum(dbscan_labels == -1)

print("DBSCAN INITIAL RESULTS")
print("-" * 35)
print("Number of clusters:", n_clusters)
print("Number of noise points:", n_noise)

print("\nFoods per cluster:")
print(
    pd.Series(dbscan_labels, name="Cluster")
    .value_counts()
    .sort_index()
)

# Calculate metrics only if there are at least 2
# non-noise clusters and enough assigned foods
non_noise_mask = dbscan_labels != -1
non_noise_labels = dbscan_labels[non_noise_mask]
n_non_noise_clusters = len(set(non_noise_labels))

if n_non_noise_clusters >= 2:
    X_assigned = np.asarray(X_dbscan)[non_noise_mask]

    print("\nPerformance metrics (excluding noise):")
    print(
        "Silhouette Score:",
        round(silhouette_score(X_assigned, non_noise_labels), 4)
    )
    print(
        "Davies-Bouldin Index:",
        round(davies_bouldin_score(X_assigned, non_noise_labels), 4)
    )
    print(
        "Calinski-Harabasz Score:",
        round(calinski_harabasz_score(X_assigned, non_noise_labels), 4)
    )
else:
    print(
        "\nNot enough non-noise clusters to calculate "
        "the clustering metrics."
    )
```

**DBSCAN has a lower Davies–Bouldin index, but it excludes almost 40% of the foods from its clusters.** 

**K-Means has a higher silhouette score and assigns every food to a cluster.** 

**Neither model is automatically the winner; their results reflect different clustering approaches.**

```python

from sklearn.cluster import DBSCAN
from sklearn.metrics import (
    silhouette_score,
    davies_bouldin_score,
    calinski_harabasz_score
)
import numpy as np
import pandas as pd

X_dbscan = X_scaled

eps_values = [0.5, 0.75, 1.0, 1.25, 1.5, 2.0, 2.5]
results = []

for eps in eps_values:
    model = DBSCAN(
        eps=eps,
        min_samples=5
    )

    labels = model.fit_predict(X_dbscan)

    noise_mask = labels == -1
    assigned_mask = ~noise_mask

    n_clusters = len(set(labels) - {-1})
    n_noise = int(noise_mask.sum())

    row = {
        "eps": eps,
        "clusters": n_clusters,
        "noise_points": n_noise,
        "noise_percent": round(n_noise / len(labels) * 100, 1),
        "silhouette": np.nan,
        "davies_bouldin": np.nan,
        "calinski_harabasz": np.nan
    }

    # Calculate metrics for assigned foods only
    assigned_labels = labels[assigned_mask]
    X_assigned = np.asarray(X_dbscan)[assigned_mask]

    if (
        n_clusters >= 2
        and len(set(assigned_labels)) >= 2
        and len(assigned_labels) > n_clusters
    ):
        row["silhouette"] = round(
            silhouette_score(X_assigned, assigned_labels), 4
        )
        row["davies_bouldin"] = round(
            davies_bouldin_score(X_assigned, assigned_labels), 4
        )
        row["calinski_harabasz"] = round(
            calinski_harabasz_score(X_assigned, assigned_labels), 4
        )

    results.append(row)

results_df = pd.DataFrame(results)
display(results_df)
```
